```python
import cmath

try:
    Kc = float(input("Enter Kc: "))
    A0 = float(input("Enter initial concentration of A: "))
    B0 = float(input("Enter initial concentration of B: "))

    if Kc <= 0:
        print("Error: Kc must be greater than 0.")
    elif A0 < 0 or B0 < 0:
        print("Error: Concentrations cannot be negative.")
    else:
        # Coefficients for a*x^2 + b*x + c = 0
        a = Kc
        b = -(Kc * (A0 + B0) + 1)
        c = Kc * A0 * B0

        print(f"a = {a}")
        print(f"b = {b}")
        print(f"c = {c}")

        discriminant = b**2 - 4*a*c
        print(f"Discriminant = {discriminant}")

        root1 = (-b + cmath.sqrt(discriminant)) / (2*a)
        root2 = (-b - cmath.sqrt(discriminant)) / (2*a)

        print(f"Root 1 = {root1}")
        print(f"Root 2 = {root2}")

except ValueError:
    print("Error: Text is not a valid number.")
```

```python
for x in (root1, root2):
    if x.imag == 0 and 0 <= x.real <= 0.5:
        physical_root = x.real
        print("Physically meaningful root:", physical_root)
        break
else:
    print("No physically meaningful root found.")
for x in (root1, root2):
    if abs(x.imag) < 1e-10 and -1e-10 <= x.real <= 0.5 + 1e-10:
        physical_root = x.real
        break
else:
    print("No physically meaningful root found.")
```

```python
# Select the valid root
A0 = float(input("Enter initial concentration of A: "))
B0 = float(input("Enter initial concentration of B: "))
physical_roots = []

for x in (root1, root2):
    if abs(x.imag) < 1e-10 and 0 <= x.real <= min(A0, B0):
        physical_roots.append(x.real)

if physical_roots:
    x = physical_roots[0]

    equilibrium_A = A0 - x
    equilibrium_B = B0 - x
    equilibrium_C = x

    limiting_reactant_initial = min(A0, B0)
    percent_conversion = (x / limiting_reactant_initial) * 100

    print(f"[A] equilibrium = {equilibrium_A:.3f} M")
    print(f"[B] equilibrium = {equilibrium_B:.3f} M")
    print(f"[C] equilibrium = {equilibrium_C:.3f} M")
    print(f"Limiting reactant conversion = {percent_conversion:.3f}%")
else:
    print("No physically meaningful equilibrium root found.")
```

```python
import math

initial_concentration = 1.0  # mol/L
rate_constant = 0.12         # 1/min
time = 5                     # minutes

# Euler method
euler_concentration = initial_concentration
for minute in range(1, time + 1):
    derivative = -rate_constant * euler_concentration
    euler_concentration += derivative * 1

# Exact value using math.exp()
exact_concentration = initial_concentration * math.exp(
    -rate_constant * time
)

# Absolute error
absolute_error = abs(exact_concentration - euler_concentration)

print(f"Exact concentration at 5 min: {exact_concentration:.6f} mol/L")
print(f"Euler concentration at 5 min: {euler_concentration:.6f} mol/L")
print(f"Absolute error: {absolute_error:.6f} mol/L")
```

```python
import math

initial_concentration = 1.0
rate_constant = 0.12
step_time = 1

time = 0
concentration = initial_concentration

while time < 100:
    conversion = ((initial_concentration - concentration) /
                  initial_concentration) * 100

    if conversion >= 70:
        break

    # Euler simulation for the next minute
    derivative = -rate_constant * concentration
    concentration += derivative * step_time
    time += step_time

conversion = ((initial_concentration - concentration) /
              initial_concentration) * 100

print(f"Time needed: {time} min")
print(f"Final concentration: {concentration:.4f} mol/L")
print(f"Final conversion: {conversion:.2f}%")
```
