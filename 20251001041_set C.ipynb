```python
import cmath

try:
    Kc = float(input("Enter Kc: "))
    A0 = float(input("Enter initial concentration of A: "))
    B0 = float(input("Enter initial concentration of B: "))

    if Kc <= 0:
        print("Error: Kc must be greater than 0.")
    elif A0 < 0 or B0 < 0:
        print("Error: Concentrations cannot be negative.")
    else:
        # Coefficients for a*x^2 + b*x + c = 0
        a = Kc
        b = -(Kc * (A0 + B0) + 1)
        c = Kc * A0 * B0

        print(f"a = {a}")
        print(f"b = {b}")
        print(f"c = {c}")

        discriminant = b**2 - 4*a*c
        print(f"Discriminant = {discriminant}")

        root1 = (-b + cmath.sqrt(discriminant)) / (2*a)
        root2 = (-b - cmath.sqrt(discriminant)) / (2*a)

        print(f"Root 1 = {root1}")
        print(f"Root 2 = {root2}")

except ValueError:
    print("Error: Text is not a valid number.")
```

```python
import math

initial_concentration = 1.0
rate_constant = 0.12
step_time = 1

time = 0
concentration = initial_concentration

while time < 100:
    conversion = ((initial_concentration - concentration) /
                  initial_concentration) * 100

    if conversion >= 70:
        break

    # Euler simulation for the next minute
    derivative = -rate_constant * concentration
    concentration += derivative * step_time
    time += step_time

conversion = ((initial_concentration - concentration) /
              initial_concentration) * 100

print(f"Time needed: {time} min")
print(f"Final concentration: {concentration:.4f} mol/L")
print(f"Final conversion: {conversion:.2f}%")
```

```python
def factorial(number):
    # Base case: 0! = 1
    if number == 0:
        return 1

    # Recursive case
    return number * factorial(number - 1)


def combination(n, r):
    if n < 0 or r < 0:
        return "Error: n and r cannot be negative."

    if r > n:
        return "Error: r cannot be greater than n."

    result = factorial(n) // (factorial(r) * factorial(n - r))
    return result


print("Factorial of 3:", factorial(3))
print("Factorial of 5:", factorial(5))

print("Ways to select 2 components from 3:",
      combination(3, 2))

print("Ways to select 2 components from 5:",
      combination(5, 2))
```

```python
def fib_iter(n):
    """Return the first n Fibonacci numbers."""
    numbers = []
    a, b = 0, 1

    for _ in range(n):
        numbers.append(a)
        a, b = b, a + b

    return numbers


def fib_rec(k):
    """Return the k-th Fibonacci number."""
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)


n = 8
iterative_numbers = fib_iter(n)

recursive_numbers = []
for k in range(n):
    recursive_numbers.append(fib_rec(k))

if iterative_numbers == recursive_numbers:
    print("Both functions agree.")
else:
    print("The functions do not agree.")

print("Reading No.\tFibonacci No.\tTime (min)")
print("-----------\t------------\t---------")

for reading, fibonacci_number in enumerate(iterative_numbers, start=1):
    time = fibonacci_number * 3
    print(f"{reading}\t\t{fibonacci_number}\t\t{time}")
```
